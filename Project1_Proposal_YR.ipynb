Airport Network Centrality and Flight Delays

### Data Source and Research Question

For this assignment, I decided to explore the relationship between airport connectivity and flight delays in the United States. I am interested in understanding whether airports with more direct flight connections experience more delays and whether this relationship differs depending on airport size.

I will use the *Reporting Carrier On-Time Performance dataset* from the U.S. Bureau of Transportation Statistics (BTS), which is publicly available through TranStats (https://www.transtats.bts.gov/). The dataset contains information about domestic flights, including origin and destination airports, departure delays, cancellations, and flight activity. For this project, I plan to use data from January 2025.

### Network Structure and Categorical Variables

In this network, each node will represent an airport, and an edge will represent a direct flight connection between two airports. Since multiple flights may operate between the same airports, I will initially treat each unique airport pair as one connection.

The main network measure I plan to investigate is degree centrality, which represents the number of direct connections an airport has relative to the total number of airports in the network.

For the categorical variable, I plan to group airports into small, medium, and large categories based on their total flight volume during the study period. This will allow me to compare degree centrality across airports of different sizes. I may also explore geographic differences using the state associated with each airport.

### Proposed Analysis

I plan to load the CSV dataset into Python using pandas and perform some initial data cleaning, including checking for missing values and identifying unique airports and routes.

Next, I will use NetworkX to construct the airport network and calculate degree centrality for each airport. I will also calculate the percentage of flights delayed by at least 15 minutes for each airport, using departure delay information and excluding canceled flights from the delay-rate calculation.

Once the network measures are calculated, I will compare degree centrality across airport-size categories and explore whether there is an association between connectivity and departure delay rates. I may also visualize the network to identify major hubs and compare their connectivity with smaller airports.

### Hypothetical Outcome

My hypothesis is that airports with higher degree centrality may experience higher rates of departure delays because they are connected to more destinations and may be more exposed to disruptions across the flight network.

However, airport size may also play an important role. Larger airports typically handle more flights, which could influence both their connectivity and delay rates. By comparing degree centrality across airport-size categories, I hope to explore whether connectivity is associated with delays beyond differences in flight volume.

Although this analysis would not establish a causal relationship, it could help identify whether network centrality is a useful indicator for predicting airport delays.
